```python
import torch
from transformers import pipeline
from datasets import load_dataset
```

```python
reviews = [
    {"text": "This movie was amazing and wonderful!", "label": 1},
    {"text": "The movie was boring and terrible.", "label": 0},
    {"text": "I really enjoyed the excellent acting.", "label": 1},
    {"text": "The film was a complete waste of time.", "label": 0},
    {"text": "Fantastic story and brilliant performances.", "label": 1}
]

print(reviews)
```

```python
from transformers import pipeline

sentiment_model = pipeline(
    "sentiment-analysis",
    model="textattack/bert-base-uncased-imdb"
)
```

```python
for review in reviews:
    result = sentiment_model(review["text"])[0]

    print("Review:", review["text"])
    print("Actual Sentiment:",
          "Positive" if review["label"] == 1 else "Negative")
    print("Predicted Sentiment:", result["label"])
    print("Confidence:", round(result["score"], 4))
    print()
```
